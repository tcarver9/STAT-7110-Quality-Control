### Q3: Briefly describe why a CUSUM and EWMA chart would be appropriate in this scenario, given the general manager's specific concern about the aging CO2 regulator. What are the limitations of these charts?

CUSUM and EWMA charts are appropriate because the main concern is a small gradual decrease in pour volume caused by the aging CO2 regulator. Both charts use information from current and previous observations which make them more sensitive to small deviations in the process mean opposed to a Shewhart chart. CUSUM helps accumulate deviations from the target as time passes and EWMA provides greater weight to recent observations while still accounting for past data.

A limitation for both charts is that they require parameters to be selected such as a reference value and decision interval for CUSUM or the smoothing parameter for EWMA. The performance of these charts is dependant on choosing these values accurately. These charts are also better for detecting changes in the process mean rather than identifying the specific cause of the change.

### Q9: Read the Phase II data into R. Using your final Phase I estimates from Q6 and the CUSUM parameters you selected in Q9, construct the Phase II CUSUM chart. Does the chart signal an out-of-control condition? If so, at which shift number?

Yes the phase2 CUSUM chart shows that shift11 is an out of control condition. The lower CUSUM crosses the decision interval which indicates evidence of a sustained downward shift in average pour volume. This is consistent with the concern that the aging CO2 regulator is gradully causing the smaller pours.

```r
## Q9 Code ##

library(readxl)

phase2 <- read_excel("C:\Users\diirt\Desktop\STAT7110\HW2\BustersSportsBar_PhaseII.xlsx")

# mean for each shift
phase2$Mean <- rowMeans(phase2[, 2:6])

phase2[, c("Shift", "Mean")]

# final phase1 estimates
mu <- 16.0548
sigma <- 0.0671
n <- 5

# SD of subgroup mean
sigma_xbar <- sigma / sqrt(n)

# CUSUM parameters
K <- 0.5
H <- 4.77

# standardize phase2 mean
z <- (phase2$Mean - mu) / sigma_xbar

# create CUSUM values
Cplus <- numeric(nrow(phase2))
Cminus <- numeric(nrow(phase2))

for(i in 1:nrow(phase2)) {

  if(i == 1) {
    Cplus[i] <- max(0, z[i] - K)
    Cminus[i] <- min(0, z[i] + K)
  } else {
    Cplus[i] <- max(0, Cplus[i-1] + z[i] - K)
    Cminus[i] <- min(0, Cminus[i-1] + z[i] + K)
  }
}

# results
data.frame(phase2$Shift, phase2$Mean, Cplus, Cminus)

# plot

plot(phase2$Shift, Cminus,
     type = "b",
     xlab = "Shift",
     ylab = "CUSUM",
     main = "Phase II CUSUM Chart")

lines(phase2$Shift, Cplus, type = "b")

abline(h = H, lty = 2)
abline(h = -H, lty = 2)
abline(h = 0)
```

### Q10: Using your final Phase I estimates from Q6 and the EWMA parameters you selected in Q8, construct the Phase II EWMA chart. Does the chart signal an out-of-control condition? If so, at which shift number?

Yes the phase2 EWMA chart identifies the same out of control condition at shift11. The EWMA falls below its lower control limit which provides evidence of a sustained downward shift in average pour volume.

```r
## Q10 Code ##

# EWMA parameters
lambda <- 0.20
L <- 2.962

# starting value
EWMA <- numeric(nrow(phase2))
EWMA[1] <- lambda * phase2$Mean[1] +
           (1 - lambda) * mu

# calculate EWMA
for(i in 2:nrow(phase2)) {
  EWMA[i] <- lambda * phase2$Mean[i] +
             (1 - lambda) * EWMA[i-1]
}

# time
t <- 1:nrow(phase2)

# EWMA control limits
UCL <- mu + L * sigma_xbar *
       sqrt((lambda / (2 - lambda)) *
       (1 - (1 - lambda)^(2*t)))

LCL <- mu - L * sigma_xbar *
       sqrt((lambda / (2 - lambda)) *
       (1 - (1 - lambda)^(2*t)))

# results
data.frame(phase2$Shift,
           phase2$Mean,
           EWMA,
           LCL,
           UCL)
# plot

plot(phase2$Shift, EWMA,
     type = "b",
     xlab = "Shift",
     ylab = "EWMA",
     main = "Phase II EWMA Chart")

lines(phase2$Shift, UCL, lty = 2)
lines(phase2$Shift, LCL, lty = 2)
abline(h = mu)
```
