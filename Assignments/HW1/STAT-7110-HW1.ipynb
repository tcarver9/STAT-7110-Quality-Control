### Q4: Read the Phase I data into R. Calculate and report the estimated process mean and standard deviation using both the $R$ and $s$ methods.
The estimated process mean for phase1 is 1.75402mm. When using the R method, the estimated process standard deviation is 0.01145mm. Using the s method, the estimated process standard deviation is 0.01154mm. Both methods produced similar estimates of the process variability.

```r
## Q4 Code ##
library(readxl)
phase1 <- read_excel("C:/Users/diirt/Downloads/FilamentDiameter_PhaseI.xlsx")
head(phase1)

# mean
phase1$Mean <- rowMeans(phase1[, 2:6])
# range
phase1$Range <- apply(phase1[, 2:6], 1, max) -
                apply(phase1[, 2:6], 1, min)
# sd
phase1$SD <- apply(phase1[, 2:6], 1, sd)

phase1

# overall
xbarbar <- mean(phase1$Mean)
Rbar <- mean(phase1$Range)
sbar <- mean(phase1$SD)

xbarbar
Rbar
sbar

# Constants for n = 5
d2 <- 2.326
c4 <- 0.9400

# process standard deviation using R
sigma_R <- Rbar / d2

# process standard deviation using s
sigma_s <- sbar / c4

sigma_R
sigma_s
```


### Q5: Construct a Phase I $R$ chart. Does the chart show evidence of statistical control? If not, identify the out-of-control subgroup(s) by number and provide a contextual explanation as to why these OOC points may have been observed, omit these points, and replot the chart until statistical control is observed.
Subgroup 19 is out of control with a range of 0.07831. So the initial R chart is not in statistical control because subgroup 19 has a range above the upper control limit. Maybe this is because of a temporary equipment problem, material inconsistency, or another specific cause that affects the process. After removing subgroup 19 and recalculating the control limits, no other groups fall outside the R chart control limits, so now the process is in statistical control.

```r
## Q5 Code ##
# R chart constants for n = 5
D3 <- 0
D4 <- 2.114

# average range
Rbar <- mean(phase1$Range)

# control limits
R_UCL <- D4 * Rbar
R_LCL <- D3 * Rbar

Rbar
R_UCL
R_LCL

plot(phase1$Subgroup, phase1$Range,
     type = "b",
     xlab = "Subgroup",
     ylab = "Range",
     main = "Phase I R Chart")

abline(h = Rbar)
abline(h = R_UCL, lty = 2)
abline(h = R_LCL, lty = 2)

phase1_R <- phase1[phase1$Subgroup != 19, ]

#recalculate
Rbar2 <- mean(phase1_R$Range)

R_UCL2 <- D4 * Rbar2
R_LCL2 <- D3 * Rbar2

Rbar2
R_UCL2
R_LCL2

#replot
plot(phase1_R$Subgroup, phase1_R$Range,
     type = "b",
     xlab = "Subgroup",
     ylab = "Range",
     main = "Revised Phase I R Chart")

abline(h = Rbar2)
abline(h = R_UCL2, lty = 2)
abline(h = R_LCL2, lty = 2)
```

### Q6: Using only the subgroups retained after Q5, construct the Phase I $\bar{X}$ chart. Does the chart show evidence of statistical control? If not, identify the out-of-control subgroup(s) by number and provide a contextual explanation as to why these OOC points may have been observed, omit these points, and replot the chart until statistical control is observed.
The X chart is initially not in statistical control because subgroup 12 has a mean above the upper control limit. Unlike subgroup 19, the observations in subgroup 12 are relatively consistent, but all of them are shifted upward. This could be the result of a temporary change in the process such as a machine setting, temperature change, or possibly a calibration issue. Something that only caused the filament diameter to increase during that specific week.

```r
## Q6 Code ##
A2 <- 0.577

# overall mean retained
xbarbar <- mean(phase1_R$Mean)

# X-bar control limits
X_UCL <- xbarbar + A2 * Rbar2
X_LCL <- xbarbar - A2 * Rbar2

xbarbar
X_UCL
X_LCL

plot(phase1_R$Subgroup, phase1_R$Mean,
     type = "b",
     xlab = "Subgroup",
     ylab = "Mean Diameter",
     main = "Phase I X-bar Chart")

abline(h = xbarbar)
abline(h = X_UCL, lty = 2)
abline(h = X_LCL, lty = 2)

# remove 12
phase1_final_R <- phase1_R[phase1_R$Subgroup != 12, ]

# recalculate
Rbar_final <- mean(phase1_final_R$Range)
xbarbar_final <- mean(phase1_final_R$Mean)

X_UCL_final <- xbarbar_final + A2 * Rbar_final
X_LCL_final <- xbarbar_final - A2 * Rbar_final

xbarbar_final
X_UCL_final
X_LCL_final

# replot
plot(phase1_final_R$Subgroup, phase1_final_R$Mean,
     type = "b",
     xlab = "Subgroup",
     ylab = "Mean Diameter",
     main = "Final Phase I X-bar Chart")

abline(h = xbarbar_final)
abline(h = X_UCL_final, lty = 2)
abline(h = X_LCL_final, lty = 2)

```
